```python
# ===============================
# Helper Functions
# ===============================

def rotl32(x, n):
    return ((x << n) & 0xffffffff) | (x >> (32 - n))


def bit_diff(a, b):
    return bin(a ^ b).count("1")


def compare_outputs(original, modified):
    diffs = []
    total = 0
    for i in range(len(original)):
        diff = bit_diff(original[i], modified[i])
        diffs.append(diff)
        total += diff
    return diffs, total


def print_outputs(label, outputs):
    print(f"\n{label}:")
    for i, val in enumerate(outputs):
        print(f"z{i} = 0x{val:08x}")


# ===============================
# Extended QR-F (Task 3)
# ===============================

def extended_qrf(x0, x1, x2, x3, x4, x5):
    z1 = x1 ^ rotl32((x0 + x5) & 0xffffffff, 2)
    z2 = x2 ^ rotl32((z1 + x3) & 0xffffffff, 4)
    z3 = x3 ^ rotl32((z2 + x4) & 0xffffffff, 7)
    z4 = x4 ^ rotl32((z3 + x0) & 0xffffffff, 8)
    z5 = x5 ^ rotl32((z4 + z1) & 0xffffffff, 12)
    z0 = x0 ^ rotl32((z5 + z2) & 0xffffffff, 16)

    return (z0, z1, z2, z3, z4, z5)


# ===============================
# Standard ChaCha Quarter Round (Task 5)
# ===============================

def chacha_qr(a, b, c, d):
    a = (a + b) & 0xffffffff
    d ^= a
    d = rotl32(d, 16)

    c = (c + d) & 0xffffffff
    b ^= c
    b = rotl32(b, 12)

    a = (a + b) & 0xffffffff
    d ^= a
    d = rotl32(d, 8)

    c = (c + d) & 0xffffffff
    b ^= c
    b = rotl32(b, 7)

    return (a, b, c, d)


# ===============================
# Modified Rotation Version (Task 6)
# ===============================

def extended_qrf_modified(x0, x1, x2, x3, x4, x5):
    # Changed rotation 16 → 15
    z1 = x1 ^ rotl32((x0 + x5) & 0xffffffff, 2)
    z2 = x2 ^ rotl32((z1 + x3) & 0xffffffff, 4)
    z3 = x3 ^ rotl32((z2 + x4) & 0xffffffff, 7)
    z4 = x4 ^ rotl32((z3 + x0) & 0xffffffff, 8)
    z5 = x5 ^ rotl32((z4 + z1) & 0xffffffff, 12)
    z0 = x0 ^ rotl32((z5 + z2) & 0xffffffff, 15)  # modified

    return (z0, z1, z2, z3, z4, z5)


# ===============================
# MAIN PROGRAM
# ===============================

def main():
    # Input values
    x0 = 0x11111111
    x1 = 0x01020304
    x2 = 0x9b8d6f43
    x3 = 0x01234567
    x4 = 0xdeadbeef
    x5 = 0xcafebabe

    print("=== ORIGINAL INPUTS ===")
    print(f"x0 = 0x{x0:08x}")
    print(f"x1 = 0x{x1:08x}")
    print(f"x2 = 0x{x2:08x}")
    print(f"x3 = 0x{x3:08x}")
    print(f"x4 = 0x{x4:08x}")
    print(f"x5 = 0x{x5:08x}")

    # ===============================
    # Task 4 — Diffusion
    # ===============================
    original = extended_qrf(x0, x1, x2, x3, x4, x5)

    x0_flip = x0 ^ 0x1
    modified = extended_qrf(x0_flip, x1, x2, x3, x4, x5)

    print_outputs("Original Extended QR-F Output", original)
    print_outputs("Modified (x0 flipped) Output", modified)

    diffs, total = compare_outputs(original, modified)

    print("\nBit Differences (Extended QR-F):")
    for i, d in enumerate(diffs):
        print(f"z{i}: {d} bits")

    print(f"Total changed bits: {total}")

    # ===============================
    # Task 5 — Compare with ChaCha20
    # ===============================
    print("\n=== ChaCha20 Comparison ===")

    orig_chacha = chacha_qr(x0, x1, x2, x3)
    mod_chacha = chacha_qr(x0_flip, x1, x2, x3)

    print_outputs("Original ChaCha QR", orig_chacha)
    print_outputs("Modified ChaCha QR", mod_chacha)

    diffs_c, total_c = compare_outputs(orig_chacha, mod_chacha)

    print("\nBit Differences (ChaCha20):")
    for i, d in enumerate(diffs_c):
        print(f"word{i}: {d} bits")

    print(f"Total changed bits (ChaCha20): {total_c}")

    # ===============================
    # Task 6 — Rotation Modification
    # ===============================
    print("\n=== Modified Rotation (16 → 15) ===")

    mod_rot = extended_qrf_modified(x0, x1, x2, x3, x4, x5)
    mod_rot_flip = extended_qrf_modified(x0_flip, x1, x2, x3, x4, x5)

    diffs_r, total_r = compare_outputs(mod_rot, mod_rot_flip)

    print_outputs("Modified Rotation Output", mod_rot)

    print("\nBit Differences (Modified Rotation):")
    for i, d in enumerate(diffs_r):
        print(f"z{i}: {d} bits")

    print(f"Total changed bits: {total_r}")

    # ===============================
    # Task 7 — Input Influence
    # ===============================
    print("\n=== Input Influence Analysis ===")

    inputs = [x0, x1, x2, x3, x4, x5]

    for i in range(6):
        test_inputs = inputs.copy()
        test_inputs[i] ^= 0x1

        modified = extended_qrf(*test_inputs)
        diffs_i, total_i = compare_outputs(original, modified)

        print(f"\nFlip in x{i}:")
        for j, d in enumerate(diffs_i):
            print(f"z{j}: {d} bits")
        print(f"Total: {total_i} bits")


if __name__ == "__main__":
    main()
```
